```python
# This code snippet is to plot a personal travel map with markers and hover text for visited places, coded with the help of chatgpt.
import pandas as pd
import folium
from geopy.geocoders import Bing
import csv

cities_list = ["Linyi","Jinan","Heze","Qingdao","Yantai","Jining","Nanjing","Suzhou","Wuxi","Zhenjiang","Hefei","Chuzhou","Huangshan","Xuzhou","Shanghai","Beijing","Chongqing","Hangzhou","Taizhou","Jinhua","Lishui","Xinyang","Wuhan","Enshi","Yichang","Guangshui","Xiangyang","Shiyan","Zhijiang","Luoyang","Binzhou","Guangzhou","Yangjiang","Shenzhen","Dongguan","Hong Kong","Qiongzhong","Lanzhou","Tianshui","Longnan","Ürümqi","Kashgar","Ordos","Erenhot","Taipei","New Taipei","Tainan","Kaohsiung","Chiayi County","Nantou County","Hsinchu City","Ulaanbaatar","Zamiin-Uud","Islamabad","Faisalabad","Gilgit","Groningen","Amsterdam","Friesland","Schiermonnikoog","Zwolle","Rotterdam","Leiden","Nijmegen","Utrecht","Eindhoven","Maastricht","The Hague","Hoorn","Bremen","Aachen","Cologne", "Ghent", "Milan","Venice","Kristiansand","Glasgow", "Brussels", "Edinburgh", "Luxembourg","Assen","Krakow","Leeuwarden","Harlingen","Eemshaven","Delfzijl","Nanchang"]

# remove duplicate items in cities_list if any
cities_list = list(set(cities_list))

# Get city coordinates
my_bing_key = 'AiFFYZiae1PbiA58R1fVsD7oPM-7oSraZCS_Gyv8mkIWPOtjeSB7vfqQa6e_cVA0'

def geocode_cities(cities, api_key):
    geolocator = Bing(api_key=api_key)
    locations = []
    
    for city in cities:
        location = geolocator.geocode(city, exactly_one=True)
        if location is not None:
            city_name = location.address.split(',')[0]
            country_name = location.raw['address']['countryRegion']
            latitude = location.latitude
            longitude = location.longitude
            locations.append((city_name, country_name, latitude, longitude))
    
    return locations

coordinates = geocode_cities(cities_list, my_bing_key)

# Output the data to a CSV file
output_file = "city_coordinates.csv"

with open(output_file, mode="w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["City", "Country", "Longitude", "Latitude"])  # Write header
    for city, country, lat, lon in coordinates:
        writer.writerow([city, country, lon, lat])
        
# Read data from the CSV file
data = pd.read_csv("city_coordinates.csv")

# Create a Folium Map
world_map = folium.Map(location=[0, 0], zoom_start=2, tiles="OpenStreetMap")

# Add Markers for each city with Tooltips
for index, row in data.iterrows():
    city = row["City"]
    country = row["Country"]
    lat = row["Latitude"]
    lon = row["Longitude"]
    tooltip_text = f"{city}, {country}"
    folium.Marker(location=[lat, lon], tooltip=tooltip_text).add_to(world_map)

# Display the map with mouse-hovering tooltips
# world_map

# # Save the map to an HTML file
world_map.save("liteng_footprint.html")
```
